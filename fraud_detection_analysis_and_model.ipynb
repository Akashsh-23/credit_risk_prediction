## Feature selection, preprocessing, and modeling

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
df_model = df.drop(columns=['nameOrig', 'nameDest', 'isFlaggedFraud'], axis=1)

categorical = ['type']
numeric = ['amount', 'oldbalanceOrg', 'newbalanceOrig',
           'oldbalanceDest', 'newbalanceDest']

y = df_model['isFraud']
X = df_model.drop(columns='isFraud', axis=1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=42
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric),
        ('cat', OneHotEncoder(drop='first'), categorical)
    ],
    remainder='drop'
)
```

```python
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(class_weight='balanced', max_iter=1000))
])
```

```python
pipeline.fit(X_train, y_train)
```

## Evaluate the model

```python
y_pred = pipeline.predict(X_test)
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
from sklearn.metrics import roc_auc_score, roc_curve

# probability of the positive class (fraud = 1)
y_proba = pipeline.predict_proba(X_test)[:, 1]

auc_score = roc_auc_score(y_test, y_proba)
print(f'AUC-ROC Score: {auc_score:.4f}')
```

```python
pipeline.score(X_test, y_test)
```
